```python
#%pip install ultralytics

from ultralytics import YOLO

model = YOLO("yolov8n.pt") # Este é um modelo pequeno, que pode ser usado para testes.

results = model("https://blog.diasbike.com.br/wp-content/uploads/2023/08/01_-_cachorro.jpg") # https://ultralytics.com/images/bus.jpg
```

```python
for result in results:
    boxes = result.boxes
    result.save(filename="result.jpg")
```

```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

img = "result.jpg"

image = mpimg.imread(img)
plt.imshow(image)
plt.axis("off")
plt.show()
```
