```python
#%pip install ultralytics

from ultralytics import YOLO

model = YOLO("yolov8n.pt") # Este é um modelo pequeno, que pode ser usado para testes.

results = model("https://www.shutterstock.com/image-vector/diverse-group-people-waiting-patiently-260nw-2659354119.jpg")
```
