# <font color="yellow">Contextualização dos modelos YOLO de Ultralytics</font>

``YOLO`` significa ``You Only Look Once`` e é uma família de modelos de visão computacional muito usada para detecção de objetos em tempo real. A ideia principal do ``YOLO`` é analisar a imagem inteira de uma só vez e prever, rapidamente, onde estão os objetos e quais são suas classes.

A biblioteca ``Ultralytics YOLO`` facilita o uso desses modelos em tarefas como:

- Detecção de objetos;
- Segmentação de imagens;
- Classificação;
- Estimativa de pose;
- Rastreamento de objetos;
- Detecção com caixas orientadas, chamada OBB.

Segundo a documentação da ``Ultralytics``, os modelos mais recentes recomendados incluem ``YOLO11`` e ``YOLO26``. O ``YOLO11`` é uma versão moderna e estável para várias tarefas de visão computacional, enquanto o ``YOLO26`` é apresentado como a geração mais recente, com foco em inferência otimizada, implantação em dispositivos de borda e execução sem NMS tradicional.


Link de estudo:

* [Ultralytics: YOLOv8](https://docs.ultralytics.com/models/yolov8/)
